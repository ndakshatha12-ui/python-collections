## Lists

```python
fruits = ["apple", "banana", "mango"]
fruits.append("grape")
fruits.remove("banana")
fruits.sort()
print(fruits)
```

```python
print(fruits[0], fruits[-1])
print(len(fruits))
```

```python
nums = [5, 2, 9, 1, 7]
nums.sort(reverse=True)
print(nums)
```

## Tuples

```python
coords = (12.9, 77.6)
print(coords)

x, y = coords
print(x, y)
```

```python
# tuples are immutable
try:
    coords[0] = 1
except TypeError as e:
    print("Error:", e)
```

## Sets

```python
tags = {"python", "sql", "python", "excel"}
tags.add("pandas")
tags.discard("excel")
print(tags)
print("sql" in tags)
```

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}
print("Union:", a.union(b))
print("Intersection:", a.intersection(b))
print("Difference:", a.difference(b))
```

## Dictionaries

```python
student = {"name": "Asha", "marks": 88, "passed": True}
student["grade"] = "A"
print(student)
```

```python
print(student.get("marks"))
print(student.keys())
print(student.values())
```

```python
student.update({"marks": 95})
print(student)
```

```python
scores = {"math": 90, "science": 85, "english": 78}
for subject, mark in scores.items():
    print(subject, mark)
```

## Comparison table

| Type | Ordered | Mutable | Duplicates allowed | Access by | Syntax |
|---|---|---|---|---|---|
| List | Yes | Yes | Yes | Index | `[1, 2, 3]` |
| Tuple | Yes | No | Yes | Index | `(1, 2, 3)` |
| Set | No | Yes | No | Membership (`in`) | `{1, 2, 3}` |
| Dictionary | Yes (insertion order) | Yes | Keys must be unique | Key | `{"a": 1}` |

## Interview questions

**What is the difference between a list, tuple, set, and dictionary?**
A list is an ordered, changeable collection that allows duplicates, accessed by index. A tuple is the same as a list but immutable — once created, its values can't change. A set is unordered and only holds unique values, with no indexing, and is optimized for membership checks and comparing groups (union, intersection). A dictionary stores key-value pairs and is accessed by key instead of position.

**Why are sets useful for finding unique values?**
A set automatically drops duplicates on creation, so converting a list to a set is a quick way to get only the distinct values. Membership checks (`x in my_set`) are also much faster on a set than on a list, since sets are hash-based.

**How do you access a dictionary value?**
With square brackets and the key, e.g. `student["marks"]`, or with `.get("marks")`, which is safer since it returns `None` (or a default) instead of raising an error if the key doesn't exist.
